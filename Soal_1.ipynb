```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Kategori": ["Elektronik", "Fesyen", "Elektronik", "Perabotan", "Fesyen",
                 "Elektronik", "Fesyen", "Perabotan", "Elektronik", "Perabotan",
                 "Fesyen", "Elektronik", "Perabotan", "Fesyen", "Elektronik"],
    "Pendapatan": [1000000, 750000, 1250000, 500000, 900000,
                   1100000, 800000, 600000, 1500000, 550000,
                   700000, 1400000, 650000, 850000, 1050000]
}

df = pd.DataFrame(data)
avg_rev = df.groupby("Kategori")["Pendapatan"].mean()

avg_rev.plot(kind="bar")
plt.title("Rata-rata Pendapatan per Kategori Produk")
plt.xlabel("Kategori Produk")
plt.ylabel("Rata-rata Pendapatan (Rp)")
plt.show()
```
